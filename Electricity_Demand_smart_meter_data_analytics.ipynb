DAY 2 ---- LAYOUUNE DATA

```python
# Select Laayoune from our combined dataset
city_data = energy_data[
    energy_data['City'] == 'Laayoune'
].copy()

# Convert timestamps to datetime
city_data['DateTime'] = pd.to_datetime(
    city_data['DateTime']
)

# Sort chronologically
city_data = city_data.sort_values(
    'DateTime'
).reset_index(drop=True)

# Define our zone columns
zones = ['zone1', 'zone2', 'zone3', 'zone4', 'zone5']

print("Total observations:", len(city_data))
print("Duplicate timestamps:",
      city_data['DateTime'].duplicated().sum())

display(city_data.head())
```

```python
# Show all rows with repeated timestamps
duplicates = city_data[
    city_data['DateTime'].duplicated(keep=False)
]

display(
    duplicates.sort_values('DateTime')
)
```

```python
# Identify all records with repeated timestamps
duplicate_records = city_data[
    city_data['DateTime'].duplicated(keep=False)
].copy()

# Summarise repeated timestamps by date
duplicate_records['Date'] = (
    duplicate_records['DateTime'].dt.date
)

duplicate_summary = (
    duplicate_records.groupby('Date')
    .agg(
        total_rows=('DateTime', 'size'),
        unique_times=('DateTime', 'nunique')
    )
)

duplicate_summary['extra_rows'] = (
    duplicate_summary['total_rows']
    - duplicate_summary['unique_times']
)

display(duplicate_summary)

# Preserve original data
city_data_original = city_data.copy()
```

```python
summary = []

for zone in zones:
    readings = city_data[zone]

    summary.append({
        'Zone': zone,
        'Mean (A)': readings.mean(),
        'Maximum (A)': readings.max(),
        'Zero Count': (readings == 0).sum(),
        'Zero (%)': (readings == 0).mean() * 100,
        '99th Percentile (A)': readings.quantile(0.99)
    })

summary_df = pd.DataFrame(summary)

display(summary_df.round(2))
```

```python
# Count how many zones record zero at each timestamp
city_data['Zero_Zones'] = (
    city_data[zones] == 0
).sum(axis=1)

# Count how often different numbers of zones record zero
zero_summary = (
    city_data['Zero_Zones']
    .value_counts()
    .sort_index()
)

display(zero_summary)
```

```python
# Show timestamps when at least 3 zones recorded zero
multiple_zero_events = city_data[
    city_data['Zero_Zones'] >= 3
]

display(
    multiple_zero_events[
        ['DateTime'] + zones + ['Zero_Zones']
    ].head(20)
)

print(
    "Rows with at least 3 zero zones:",
    len(multiple_zero_events)
)
```

```python
# Work with the data in chronological order
events_data = city_data.sort_values('DateTime').copy()

# Identify observations with at least 3 zones at zero
events_data['Multiple_Zero'] = (
    events_data['Zero_Zones'] >= 3
)

# Identify a new event when:
# 1. The previous row was not a multiple-zero observation
# 2. Or the timestamp gap is not 10 minutes

new_event = (
    events_data['Multiple_Zero']
    & (
        ~events_data['Multiple_Zero'].shift(fill_value=False)
        | events_data['DateTime'].diff().ne(
            pd.Timedelta(minutes=10)
        )
    )
)

# Assign event numbers
events_data['Event_ID'] = new_event.cumsum()

# Keep only multiple-zero observations
zero_events = events_data[
    events_data['Multiple_Zero']
].copy()

# Summarise each event
event_summary = zero_events.groupby('Event_ID').agg(
    Start=('DateTime', 'min'),
    End=('DateTime', 'max'),
    Observations=('DateTime', 'size'),
    Maximum_Zones_Affected=('Zero_Zones', 'max')
)

# Calculate the span from first to last zero sample
event_summary['Observed_Span_Minutes'] = (
    event_summary['End'] - event_summary['Start']
).dt.total_seconds() / 60

display(event_summary)
```

```python
# Select Event 4
event4 = event_summary.loc[4]

start = event4['Start'] - pd.Timedelta(hours=3)
end = event4['End'] + pd.Timedelta(hours=3)

# Extract observations around the event
event4_data = city_data[
    city_data['DateTime'].between(start, end)
].copy()

# Plot all five zones
plt.figure(figsize=(15, 6))

for zone in zones:
    plt.plot(
        event4_data['DateTime'],
        event4_data[zone],
        label=zone
    )

plt.axvspan(
    event4['Start'],
    event4['End'],
    color='red',
    alpha=0.12,
    label='Multiple-zero event'
)

plt.title('Investigation of Event 4: Simultaneous Zero-Current Readings')
plt.xlabel('Date and Time')
plt.ylabel('Electrical Current (A)')
plt.legend()
plt.grid(alpha=0.3)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Select all observations belonging to Event 4
event4_records = zero_events[
    zero_events['Event_ID'] == 4
].copy()

# Count zero readings by zone during Event 4
zero_counts = (
    event4_records[zones] == 0
).sum()

print("Zero readings by zone during Event 4:")
display(zero_counts)

# Check the first and last observations
display(
    event4_records[
        ['DateTime'] + zones
    ].head(5)
)

display(
    event4_records[
        ['DateTime'] + zones
    ].tail(5)
)
```

```python
# Inspect variation in each zone during Event 4
event4_stats = event4_records[zones].agg(
    ['min', 'max', 'mean', 'std', 'nunique']
).T

event4_stats.columns = [
    'Minimum',
    'Maximum',
    'Mean',
    'Std Deviation',
    'Unique Values'
]

display(event4_stats.round(4))
```

```python
# Count distinct values and measure variation during Event 4
event4_stats = event4_records[zones].agg(
    ['min', 'max', 'std', 'nunique']
).T

event4_stats.columns = [
    'Minimum',
    'Maximum',
    'Standard Deviation',
    'Unique Values'
]

display(event4_stats.round(6))
```

```python
# Preserve the original working dataset
city_data_clean = city_data.copy()

# Create a flag for the suspected anomaly
city_data_clean['Suspected_Data_Anomaly'] = False

# Flag Event 4 using its start and end timestamps
event4_start = event_summary.loc[4, 'Start']
event4_end = event_summary.loc[4, 'End']

event4_mask = city_data_clean['DateTime'].between(
    event4_start,
    event4_end
)

city_data_clean.loc[
    event4_mask, 'Suspected_Data_Anomaly'
] = True

print(
    "Flagged observations:",
    city_data_clean['Suspected_Data_Anomaly'].sum()
)

# Create a separate dataset for initial demand-pattern analysis
normal_data = city_data_clean[
    ~city_data_clean['Suspected_Data_Anomaly']
].copy()

print("Original observations:", len(city_data_clean))
print("Remaining observations:", len(normal_data))
```

Day 3 --- Code

```python
# Calculate average current between midnight and 05:59
night_data = daily_data[
    daily_data['Hour'].between(0, 5)
]

daily_features['Night_Mean'] = (
    night_data.groupby('Date')['zone3'].mean()
)

# Confirm that the column exists
print(daily_features.columns.tolist())

display(daily_features.head())
```

```python
# Identify dates affected by Events 1, 2 and 3
other_event_dates = set()

for event_id in [1, 2, 3]:
    event_rows = zero_events[
        zero_events['Event_ID'] == event_id
    ]

    other_event_dates.update(
        event_rows['DateTime'].dt.date.unique()
    )

print("Other flagged event dates:")
print(sorted(other_event_dates))

# Prepare the daily feature dataset
feature_columns = [
    'Mean_Current',
    'Max_Current',
    'Min_Current',
    'Std_Current',
    'Morning_Mean',
    'Afternoon_Mean',
    'Evening_Mean',
    'Night_Mean'
]

valid_days = daily_features[
    (daily_features['Observations'] == 144) &
    (daily_features['Unique_Timestamps'] == 144)
].copy()

# Exclude dates affected by Events 1–3
valid_days = valid_days[
    ~valid_days.index.isin(other_event_dates)
]

# Remove days with missing feature values
valid_days = valid_days.dropna(
    subset=feature_columns
)

print("Days ready for clustering:", len(valid_days))

display(valid_days[feature_columns].head())
```

```python
from sklearn.preprocessing import StandardScaler

# Select our eight daily features
X = valid_days[feature_columns].copy()

# Create the scaler
scaler = StandardScaler()

# Fit and transform the features
X_scaled = scaler.fit_transform(X)

# Convert back into a DataFrame
X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=feature_columns,
    index=X.index
)

print("Original features:")
display(X.head())

print("Scaled features:")
display(X_scaled_df.head())
```

check if scaling worked

```python
# Check the mean and standard deviation
scaling_check = pd.DataFrame({
    'Mean': X_scaled_df.mean(),
    'Standard Deviation': X_scaled_df.std(ddof=0)
})

display(scaling_check.round(4))
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

k_values = range(2, 11)

inertias = []
silhouette_scores = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = model.fit_predict(X_scaled_df)

    inertias.append(model.inertia_)

    silhouette_scores.append(
        silhouette_score(X_scaled_df, labels)
    )

results = pd.DataFrame({
    'K': list(k_values),
    'Inertia': inertias,
    'Silhouette_Score': silhouette_scores
})

display(results.round(4))
```

```python
plt.figure(figsize=(8, 5))
plt.plot(results['K'], results['Inertia'], marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method — Zone 3 Daily Current')
plt.xticks(list(k_values))
plt.grid(alpha=0.3)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(
    results['K'],
    results['Silhouette_Score'],
    marker='o'
)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Average Silhouette Score')
plt.title('Silhouette Analysis — Zone 3')
plt.xticks(list(k_values))
plt.grid(alpha=0.3)
plt.show()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=6,
    random_state=42,
    n_init=20
)

cluster_labels = kmeans.fit_predict(X_scaled_df)

# Preserve original current values for interpretation
clustered_days = valid_days.copy()

clustered_days['Cluster'] = cluster_labels

print("Number of days in each cluster:")
print(
    clustered_days['Cluster']
    .value_counts()
    .sort_index()
)
```

```python
# Average feature values for each cluster
cluster_summary = (
    clustered_days
    .groupby('Cluster')[feature_columns]
    .mean()
    .round(2)
)

display(cluster_summary)
```

```python
# Sort chronologically
timeline = final_clusters.sort_index().copy()

# Ensure the index contains datetime values
timeline.index = pd.to_datetime(timeline.index)

# Check gaps between dates
timeline['Day_Gap'] = (
    timeline.index.to_series().diff().dt.days
)

# Identify new runs when the cluster changes
# or when there is a missing calendar day
new_run = (
    timeline['Cluster'].ne(timeline['Cluster'].shift())
    | timeline['Day_Gap'].ne(1)
)

timeline['Run_ID'] = new_run.cumsum()

# Summarise continuous runs
run_summary = (
    timeline.groupby('Run_ID')
    .agg(
        Cluster=('Cluster', 'first'),
        Start=('Cluster', lambda x: x.index.min()),
        End=('Cluster', lambda x: x.index.max()),
        Duration_Days=('Cluster', 'size')
    )
)

print("Longest continuous runs:")
display(
    run_summary.sort_values(
        'Duration_Days',
        ascending=False
    ).head(10)
)

print("Run statistics by cluster:")
display(
    run_summary.groupby('Cluster')['Duration_Days']
    .agg(['count', 'mean', 'median', 'max'])
    .round(2)
)

# Count actual transitions between consecutive days
transitions = timeline[
    (timeline['Day_Gap'] == 1)
    & timeline['Cluster'].ne(timeline['Cluster'].shift())
]

print("Total consecutive-day cluster transitions:", len(transitions))
```
